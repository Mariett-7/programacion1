```python
import numpy as np
import pandas as pd


PRUEBA_A = np.random.normal(loc=7.2, scale=0.4, size=59)
PRUEBA_B = np.random.gamma(shape=306, scale=0.0229, size=59)
```

```python
calificaciones = pd.DataFrame({
    "A": PRUEBA_A,
    "B": PRUEBA_B
})
calificaciones
```

```python
calificaciones.describe()
```

para la prueba A, las calificaciones tienen una media de 7.130323, mientras que para la prueba B este valor es de 6.968233, lo que indica que el promedio de calificaciones es mayor en la prueba A. aún asi su desviación estandar no estan lejana siendo 0.372849 y 0.375741 para la prueba A y B respectivamente. por otro lado se tiene que la calificación minima para cada prueba fue de 6.299386 y 5.951341 lo que marca una diferencia de 0.348044 entre la nota minima de las 2 pruebas. por el momento estos datos nos muestran que si hay una diferencia promedio entre las calificaciones de ambas pruebas.

```python
from scipy import stats
```

```python
from scipy.stats import shapiro
```

```python
shapiro(calificaciones["A"])
```

```python
shapiro(calificaciones["B"])
```

como en ambas pruebas el valor p es mayor a 0.05 no se rechaza con una confianza del 95% el supuesto de normalidad

```python
stats.ttest_rel
```

```python
t_stat, p_val = stats.ttest_ind(calificaciones["A"], calificaciones["B"])
t_stat, p_val
```
