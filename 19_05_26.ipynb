```python
import re

def validar_nombre(nombre):
    if not re.search(r"^[a-zA-Z]+$", nombre):
        raise ValueError("el nombre solo debe contener letras")

def validar_edad(edad):
    if not re.match(r"^\d{3}$", edad):
        raise ValueError("la edad debe tener exactamente 3 dígitos")

def generar_contrasena(nombre, edad):
    prefijo = re.match(r"[a-zA-Z]{1,3}", nombre).group()
    return prefijo + edad + "@2026"

try:
    nombre = input("ingrese su nombre y apellido sin espacios: ").lower()
    edad = input("ingrese su edad (3 dígitos): ")

    validar_nombre(nombre)
    validar_edad(edad)

    print("contraseña generada:", generar_contrasena(nombre, edad))

except ValueError as e:
    print("Error:", e)
```
