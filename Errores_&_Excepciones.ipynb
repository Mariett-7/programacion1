```python
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}

try:
    usuario = input("Usuario: ").upper()
    password = input("Contraseña: ")

    if usuario not in usuarios:
        raise KeyError("El usuario no existe")

    if usuarios[usuario] != password:
        raise ValueError("Contraseña incorrecta")

    else:
        print("Bienvenido,", usuario)

except KeyError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

Una expresion regular es un secuencia de caracteres que define un patrón de busqueda
Se utilizan para :


*  Extraer datos
*  Reemplazar palabras
*  Limpiar cadenas de textos

el metodo search permite encontratr los indices de la primera vez que aparece un elemento dentro de mi cadena de texto

```python
import re
texto = "tengo 2 conejos y 12 gato"
resultado = re.findall(r"\d+", texto)
print(resultado)
```

el metodo findall permite encontrar los digitos. no determina cadenas, sino solo digitos

```python
import re
texto= "python es divertido"
resultado= re.findall(r"\w+", texto)
print(resultado)
```

findall tambien permite encontrat palabras completas con (\w+)

```python
import re
texto ="Universidad"
resultado= re.findall(r"[aeiou]", texto)
print(resultado)
```

```python
import re
texto= "me gusta java"
nuevo = re.sub("java", "python", texto)
print(nuevo)
```

```python
import re
texto = "Python Es Genial"
resultado = re.findall(r"[A-Z]", texto)
print(resultado)
```

```python
import re
texto = "Hola mundo Python"
resultado = re.findall(r"\s", texto)
print(resultado)
```

El argumento \s representa espacios.

```python
import re
texto = "Tengo 15 libros y 200 hojas"
resultado = re.findall(r"\d+", texto)
print(resultado)
```


\d+ encuentra números completos.

```python
import re
texto = "Correos: ana@gmail.com y juan@hotmail.com"
resultado = re.findall(r"\w+@\w+\.com", texto)
print(resultado)
```

Hacer una búsqueda sencilla de correos electrónicos.

```python
import re
telefono = "3104567890"
resultado = re.fullmatch(r"\d{10}", telefono)
print(resultado)
```

fullmatch valida toda la cadena. En este caso que se tengan 10 dígitos.

```python
import re
texto = "color colour"
resultado = re.findall(r"colou?r", texto)
print(resultado)
```

Resumen rápido

\d → dígitos

\w → letras y números

\s → espacios

. → cualquier carácter

 + → una o más veces

 * → cero o más veces

? → opcional

[abc] → a, b o c

[^abc] → excepto a, b o c
